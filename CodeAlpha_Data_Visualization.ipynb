The objective of this project is to represent student performance data using charts and graphs to make the data easier to understand.

## 2. Bar Chart

A bar chart was created to compare the average scores in mathematics, reading and writing. The chart helps identify which subject has the highest and lowest average scores.

## 3. Pie Chart

A pie chart was created to show the percentage distribution of students by gender.

## 4. Score Distribution

A graph was created to compare the distribution of scores across the three subjects and understand the patterns in student performance.

## 5. Gender Comparison

A grouped bar chart was created to compare the average mathematics, reading and writing scores by gender.

## 6. Insights

The charts help compare average scores, understand the distribution of students and identify differences in subject performance. The specific findings are based on the visualizations generated from the dataset.

## 7. Conclusion

Data visualization makes it easier to identify patterns and communicate information clearly. This project uses Python, Pandas, Matplotlib and Seaborn to visualize student performance data.


```python
# Import libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python

df = pd.read_csv('/content/StudentsPerformance.csv')

print(df.head())
```

```python

scores = ['math score', 'reading score', 'writing score']
average_scores = df[scores].mean()

plt.figure(figsize=(8, 5))
average_scores.plot(kind='bar')

plt.title('Average Scores by Subject')
plt.xlabel('Subject')
plt.ylabel('Average Score')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

gender_counts = df['gender'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Student Gender Distribution')
plt.show()
```

```python

plt.figure(figsize=(9, 5))

for subject in ['math score', 'reading score', 'writing score']:
    sns.kdeplot(data=df, x=subject, label=subject)

plt.title('Distribution of Scores by Subject')
plt.xlabel('Score')
plt.ylabel('Density')
plt.legend()
plt.tight_layout()
plt.show()
```

```python

gender_scores = df.groupby('gender')[
    ['math score', 'reading score', 'writing score']
].mean()

gender_scores.plot(kind='bar', figsize=(9, 5))

plt.title('Average Exam Scores by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Score')
plt.xticks(rotation=0)
plt.legend(title='Subject')
plt.tight_layout()
plt.show()
```
